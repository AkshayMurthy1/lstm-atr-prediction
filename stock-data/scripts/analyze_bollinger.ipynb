```python
import pandas as pd
from pathlib import Path

def find_max_combo_in_df(df: pd.DataFrame):
    # Clean up potential unnamed index column and whitespace
    if df.columns[0].lower().startswith('unnamed'):
        df = df.drop(columns=[df.columns[0]])
    df.columns = [str(c).strip() for c in df.columns]

    if 'metric' not in df.columns:
        raise ValueError("Expected a 'metric' column in the table.")

    # Factor columns are everything except 'metric'
    factor_cols = [c for c in df.columns if c != 'metric']

    # Set metric as index and make sure factor columns are numeric
    tbl = df.set_index('metric')[factor_cols].apply(pd.to_numeric, errors='coerce')

    # Stack to long form and find the global max
    stacked = tbl.stack(dropna=True)
    if stacked.empty:
        return None

    max_idx = stacked.idxmax()      # ('metric_name', 'factor')
    max_val = stacked.loc[max_idx]  # numeric max value
    return max_idx[0], max_idx[1], float(max_val)

def analyze_tables(paths):
    results = {}
    for p in paths:
        p = Path(p)
        df = pd.read_csv(p)
        out = find_max_combo_in_df(df)
        if out is None:
            results[p.name] = {"metric": None, "factor": None, "value": None}
        else:
            metric, factor, value = out
            results[p.name] = {"metric": metric, "factor": factor, "value": value}
    return results
```

```python
import os
```

```python
num = 50007
root = f'../output/runtest_bollinger_{num}'
paths_1 = os.listdir(root)
```

```python
for etf in paths_1:
    paths_2 = os.listdir(os.path.join(root,etf))
    for period in paths_2:
        root_f = os.path.join(root,etf,period)
        for file in os.listdir(root_f):
            if "trades" in file:
                continue
            print(analyze_tables([os.path.join(root_f,file)]))
```
